## Word Count Program on Small Dataset

This implementation of word count is being used as a test to see how I can process the `NameAgeData.csv` file, the output the sum of names.

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("WordCount").getOrCreate()
```

```python
df = spark.read.csv("data/NameAgeData.csv", header=True)
df.show()
```

```python
name_count = df.groupBy("Name").count() # Can switch "Name" with "Age" to count ages instead
name_count.show()
```
